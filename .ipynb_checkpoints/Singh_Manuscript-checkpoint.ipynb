```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
import seaborn as sns
import os, sys, re, warnings
myList = [[r'PythonHelper']]
myList.append(sys.path[1:])
myList = [aPath for aPart in myList for aPath in aPart]
sys.path = myList
import LASSI_2 as lassi
import BinTrjProc as BR
import TrjProc as TP
warnings.filterwarnings("ignore")

def flatten_2d_list(lst):
    return [item for sublist in lst for item in sublist]

def distance_ij(p1,p2,boxSize):
    halfBox=boxSize/2
    d=p2-p1
    d_x,d_y,d_z=[np.abs(ele) for ele in d]
    D_x=d_x if d_x<=halfBox else boxSize-d_x
    D_y=d_y if d_y<=halfBox else boxSize-d_y
    D_z=d_z if d_z<=halfBox else boxSize-d_z

    distance_wa=np.sqrt(D_x*D_x+D_y*D_y+D_z*D_z)
    
    return(distance_wa)

def center_of_mass(positions, box):
    N_p=len(positions) 
    center_vector=[box[0][1]-box[0][0],box[1][1]-box[1][0],box[2][1]-box[2][0]]
    com=np.zeros(3)
    
    for dim in range(3):
        pos_dim=positions[:,dim]
        theta_dim=[(ele+center_vector[dim]/2)*np.pi*2/(center_vector[dim]) for ele in pos_dim]
        eta_dim=[np.cos(ele) for ele in theta_dim]
        zeta_dim=[np.sin(ele) for ele in theta_dim]
        mean_eta=np.mean(eta_dim)
        mean_zeta=np.mean(zeta_dim)

        mean_theta=np.arctan2(-mean_zeta,-mean_eta)+np.pi
        com[dim]=center_vector[dim]*mean_theta/2/np.pi
        com[dim]-=(center_vector[dim]/2)

        if (com[dim]<box[dim][0]):
            com[dim]+=center_vector[dim]
            
    return(com)
    
def beads5_positions(chains_in_cluster, coords_chains, a_end, b_end, c_end, d_end, e_end, chain_length):
    As=[coords_chains[(ele-1)*chain_length:(ele-1)*chain_length+a_end].tolist() for ele in chains_in_cluster]
    Bs=[coords_chains[(ele-1)*chain_length+a_end:(ele-1)*chain_length+b_end].tolist() for ele in chains_in_cluster]
    Cs=[coords_chains[(ele-1)*chain_length+b_end:(ele-1)*chain_length+c_end].tolist() for ele in chains_in_cluster]
    Ds=[coords_chains[(ele-1)*chain_length+c_end:(ele-1)*chain_length+d_end].tolist() for ele in chains_in_cluster]
    Es=[coords_chains[(ele-1)*chain_length+d_end:(ele-1)*chain_length+e_end].tolist() for ele in chains_in_cluster]
    return As, Bs, Cs, Ds, Es
```

```python
bead_colors = ['red','goldenrod','limegreen','blue','deeppink']
rna_colors = ['skyblue','cornflowerblue','blue'] 

# G3BP1 coarse-grained structure
bead_labels =['NTF2','IDR1','IDR2','RRM1','IDR3']  
a_end, b_end, c_end, d_end, e_end = 2,10,16,18,22
chain1_length = e_end
no_chain1 = 2500

# Other
box_L=120
radial_bins = np.arange(0,120.25,0.25)
```

## For a scaling factor, compute cluster size distribution and domain-resolved radial density profile - Figure 6B-C

```python
scaling_factor = '1.25'
all_coords = np.load('g3bp1_data/{c}_G3BP1_120_22_coords.npy'.format(c=scaling_factor), allow_pickle=True)        
all_clusters = np.load('g3bp1_data/{c}_G3BP1_120_22_cluster_ids.npy'.format(c=scaling_factor), allow_pickle=True)        
all_clusterSizes = np.load('g3bp1_data/{c}_G3BP1_120_22_cluster_sizes.npy'.format(c=scaling_factor), allow_pickle=True)        
```

```python
# Compute cluster size distribution
clusterSize_hist = []
for snapshot in range(0,len(all_clusterSizes)):
    curr_sizeList = all_clusterSizes[snapshot]
    All_hist, r = np.histogram(curr_sizeList, bins=np.arange(1,no_chain1+1,1), density=True)
    clusterSize_hist.append(All_hist)
```

```python
# Compute density of each bead type at radial bins from center of mass of clusters
print('Number of bead types:', len(bead_labels))
distances_from_com = [[] for i in range(0,len(bead_labels))]

minSize = 10
maxSize = 100

for snapshot in range(0,50):
    
    for cluster in range(0,len(all_clusterSizes[snapshot])):        
        interest = all_clusters[snapshot][cluster]

        if len(interest) >= minSize and len(interest) <= maxSize:
            all_pos=[all_coords[snapshot][(ele-1)*chain1_length:ele*chain1_length].tolist() for ele in interest]
            points = np.array([coord for coord in flatten_2d_list(all_pos)])

            # compute center of mass
            com=center_of_mass(points, [[0,box_L],[0,box_L],[0,box_L]])

            # determine positions of each bead type
            pos = beads5_positions(interest, all_coords[snapshot], a_end, b_end, c_end, d_end, e_end, chain1_length)

            # compute distances for each bead from com
            for domain in range(0,len(bead_labels)):
                tmp_dist = []
                tmp_points = np.array([coord for coord in flatten_2d_list(pos[domain])])
                for bead in range(0,len(tmp_points)):
                    tmp_dist.append(distance_ij(tmp_points[bead],com,box_L))

                All_hist, r = np.histogram(tmp_dist, bins = radial_bins, density=True)
                distances_from_com[domain].append(All_hist)    
```

```python
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(5, 7), dpi=150, sharex=False, gridspec_kw={'height_ratios': [1, 1.5]})
fig.tight_layout(pad=4)
axes_figs = [ax1, ax2]

x =  np.arange(1,no_chain1,1)
y = np.mean(clusterSize_hist, axis=0)

axes_figs[0].plot(x, y, marker='o', linestyle='', markersize=2, color='black')
axes_figs[0].set_xscale('log')
axes_figs[0].set_yscale('log')
axes_figs[0].set_ylabel(r'$P(n)$', rotation=90)
axes_figs[0].set_xlabel('Cluster Size $n$')
axes_figs[0].set_xticks([1,10,100,1000,3000])    
axes_figs[0].set_xlim(0, 3000)
axes_figs[0].set_ylim(0.001, 2)
axes_figs[0].axvspan(minSize, maxSize, color='pink', alpha=0.25, zorder=0)
axes_figs[0].set_title(r'G3BP1 only $s=$'+str(scaling_factor))

max_range=150
for domain in range(0, len(pos)):
    axes_figs[1].plot(radial_bins[0:max_range], np.mean(distances_from_com[domain], axis=0)[0:max_range], 
             linestyle='-', color=bead_colors[domain], label=bead_labels[domain], linewidth=1.5)
    mean_vals = np.mean(distances_from_com[domain], axis=0)[0:max_range]
    std_vals  = np.std(distances_from_com[domain], axis=0)[0:max_range]

axes_figs[1].legend(fontsize=8, loc='upper right', title_fontsize=9, title='G3BP1 Domain', fancybox=True)
axes_figs[1].set_xlabel('Distance from the COM (l.u)')
axes_figs[1].set_ylabel('Density of Beads')
axes_figs[1].axhline(0, linestyle='--', c='gray', linewidth=2, alpha=0.5)
```

## Projections of G3BP1-RNA clusters - Figure 6I

```python
import pandas as pd
from scipy.stats.kde import gaussian_kde
from matplotlib.gridspec import GridSpec    

def contour_density_plot_biggest(points, axis1, axis2, chain2_length):

    x = points[:,0]
    y = points[:,1]
    z = points[:,2]

    coordinates = pd.DataFrame({'x': x, 'y': y, 'z': z})
    levels = np.linspace(0, (22+chain2_length)/np.shape(points)[0], num=10)  

    cmap = plt.cm.RdPu
    colors = cmap(np.linspace(0, 1, cmap.N))
    darker_color = np.array([0.1, 0.1, 0.1, 1])
    colors = np.vstack([colors, darker_color]) 
    extended_cmap = mcolors.ListedColormap(colors)
    norm = mcolors.BoundaryNorm(boundaries=np.append(levels, 0.01), ncolors=extended_cmap.N)
    
    xy_points = coordinates[[axis1, axis2]].values.T
    kde = gaussian_kde(xy_points, bw_method=0.02)
    x_grid, y_grid = np.arange(0, 121, 1), np.arange(0, 121, 1)
    X, Y = np.meshgrid(x_grid, y_grid)
    Z = kde(np.vstack([X.ravel(), Y.ravel()])).reshape(X.shape)
    
    return X,Y,Z,levels
```

```python
scaling_factors = ['1.25']
clusters_list = [-4,-10,-10]

for s in range(0,len(scaling_factors)):
    for r in range(0,len(rna_names)):
        curr_system = rna_names[r]
        save_name = scaling_factors[s]+'_'+curr_system+'_'+str(rna_chain_nums[r])

        no_chain2 = rna_chain_nums[r]
        no_chains = no_chain1 + no_chain2
        print(save_name)

        chain2_length = rna_chain_lengths[r]
        chain1_bound=[0, chain1_length*no_chain1]
        chain2_bound=[chain1_bound[1], chain1_bound[1]+(chain2_length*int(no_chain2))]
        
        A0, A1 =[0, chain1_length*no_chain1]
        B0, B1 =[chain1_bound[1], chain1_bound[1]+(chain2_length*int(no_chain2))]

        all_coords = np.load('g3bp1_rna_data/'+save_name+'_all_coords.npy', allow_pickle=True)        
        clusterList = np.load('g3bp1_rna_data/'+save_name+'_cluster_ids.npy', allow_pickle=True)        
        clusterSizes = np.load('g3bp1_rna_data/'+save_name+'_cluster_sizes.npy', allow_pickle=True)        

        ####### Determine G3BP1 and RNA chains in isolated cluster #######################
        frame = 49
        coord_list = all_coords[frame]
        tmp_list = clusterList[frame][np.argsort(clusterSizes[frame])[clusters_list[r]]]

        g3bp1_list = [chain for chain in tmp_list if 1 <= chain <= no_chain1]
        g3bp1_chains = []
        for ele in g3bp1_list:
            start = A0 + (ele-1)*chain1_length
            end = start + chain1_length
            g3bp1_chains.append(coord_list[start:end])
        g3bp1_points = np.array([c for chain in g3bp1_chains for c in chain])

        rna_list = [chain for chain in tmp_list if chain > no_chain1]
        rna_chains = []
        for ele in rna_list:
            b_idx = ele - (no_chain1 + 1) 
            start = B0 + b_idx*chain2_length
            end = start + chain2_length
            rna_chains.append(coord_list[start:end])
        rna_points = np.array([c for chain in rna_chains for c in chain])  

        all_points = np.concatenate((g3bp1_points, rna_points), axis=0)
        
        ####### Plot density contours #######################
        fig = plt.figure(figsize=(8, 8), dpi=150)
        gs = GridSpec(2, 2, width_ratios=[4, 1], height_ratios=[1, 4], hspace=0.1, wspace=0.1)

        ax_main  = fig.add_subplot(gs[1, 0])
        ax_top   = fig.add_subplot(gs[0, 0], sharex=ax_main)
        ax_right = fig.add_subplot(gs[1, 1], sharey=ax_main)

        X, Y, Z, levels = contour_density_plot_biggest(all_points, 'x', 'y', chain1_length)
        im = ax_main.contourf(X, Y, Z, levels=levels, cmap='inferno', extend='max')

        ax_main.set_xlabel('X')
        ax_main.set_ylabel('Y')

        ####### Compute line scan densities for G3BP1 and RNA #######################
        ####### Plot G3BP1 line scan ###################################
        pos = beads5_positions(g3bp1_list, coord_list, a_end=2, b_end=10, c_end=16, d_end=18, e_end=22, chain_length=22)
        beadCounts_top = np.zeros((len(bead_labels), 120))
        for bead in range(0, len(pos)):
            flattened_coordinates = [coord for coord in flatten_2d_list(pos[bead])]
            points = np.array(flattened_coordinates)
            count = 0
            for i in range(0, len(points)):
                if points[i, 2] >= 0:
                    beadCounts_top[bead, points[i, 0]] += 1
                    count += 1
            if count > 0:
                beadCounts_top[bead, :] *= 1 / count

        for i in range(0, len(pos)):
            ax_top.plot(np.arange(0, 120, 1), beadCounts_top[i, :], color=bead_colors[i])
        ax_top.tick_params(labelbottom=False, labelleft=True)
        ax_top.set_ylabel("G3BP1 Intensity", fontsize=10)
        ax_top.spines['right'].set_visible(False)
        ax_top.spines['top'].set_visible(False)

        ####### Plot RNA line scan ###################################
        beadCounts_right = np.zeros((1, 120))
        count = 0
        for i in range(0, len(rna_points)):
            if rna_points[i, 2] >= 0:
                beadCounts_right[0, rna_points[i, 1]] += 1
                count += 1
        if count > 0:
            beadCounts_right[0, :] *= 1 / count

        ax_right.plot(beadCounts_right[0, :], np.arange(0, 120, 1), color=rna_colors[r])
        ax_right.tick_params(labelleft=False, labelbottom=True)
        ax_right.set_xlabel("RNA Intensity", fontsize=10)
        ax_right.spines['right'].set_visible(False)
        ax_right.spines['top'].set_visible(False)
     
        plt.show()
```

```python
scaling_factors = ['1.25'] #['0.75','1.00','1.25']
rna_names = ['rA40','shortA','longA']
rna_label_names = ['rA40','Short Poly(rA)','Long Poly(rA)']
rna_chain_lengths = [6, 20, 66]
rna_chain_nums = [2750, 825, 250]

for s in range(0,len(scaling_factors)):

    for r in range(0,len(rna_names)):
        
        fig, ax = plt.subplots(figsize=(5,4),dpi=150)
        allclusterSizes = np.load('g3bp1_data/{c}_G3BP1_120_22_cluster_sizes.npy'.format(c=scaling_factors[s]), allow_pickle=True)        
    
        clusterSize_hist = []
        for snapshot in range(0,len(allclusterSizes)):
            curr_sizeList = allclusterSizes[snapshot]
            All_hist,b=np.histogram(curr_sizeList,bins=np.arange(1,2501,1), density=True)
            clusterSize_hist.append(All_hist)
    
        y = np.mean(clusterSize_hist, axis=0)    
        ax.loglog(np.arange(1,2500,1), y, marker='o',linestyle='',markersize=6,color='mediumorchid',alpha=0.75, label='G3BP1 only')
        ax.set_ylabel(r'$P(n)$', rotation=90, fontsize=15)
        
        curr_system = rna_names[r]
        save_name = scaling_factors[s]+'_'+curr_system+'_'+str(rna_chain_nums[r])

        no_chain2 = rna_chain_nums[r]
        no_chains = no_chain1 + no_chain2
        print(save_name)

        chain2_length = rna_chain_lengths[r]
        chain1_bound=[0, chain1_length*no_chain1]
        chain2_bound=[chain1_bound[1], chain1_bound[1]+(chain2_length*int(no_chain2))]

        all_coords = np.load('g3bp1_rna_data/'+save_name+'_all_coords.npy', allow_pickle=True)        
        clusterList = np.load('g3bp1_rna_data/'+save_name+'_cluster_ids.npy', allow_pickle=True)        
        clusterSizes = np.load('g3bp1_rna_data/'+save_name+'_cluster_sizes.npy', allow_pickle=True)        

        g3bp1Size_hist = []
        rnaSize_hist = []
        for snapshot in range(0,len(clusterSizes)):
            g3bp1_sizeList = []
            rna_sizeList = [] 
            
            for cc in range(0,len(clusterList[snapshot])):
                g3bp1_list = [chain for chain in clusterList[snapshot][cc] if 1 <= chain <= no_chain1]
                rna_list = [chain for chain in clusterList[snapshot][cc] if no_chain1+1 <= chain <= no_chains+1]

                g3bp1_sizeList.append(len(g3bp1_list))
                rna_sizeList.append(len(rna_list))

            All_hist,b=np.histogram(g3bp1_sizeList, bins=np.arange(1,2501,1), density=True)
            g3bp1Size_hist.append(All_hist)
            
            All_hist,b=np.histogram(rna_sizeList, bins=np.arange(1,no_chain2+1,1), density=True)
            rnaSize_hist.append(All_hist)

        x =  np.arange(1,no_chain1,1)
        y = np.mean(g3bp1Size_hist, axis=0)    
        ax.loglog(x, y, marker='o', linestyle='', markersize=6, color=rna_colors[r], alpha=0.75, label='G3BP1+'+rna_label_names[r])
        
        ax.set_xlabel('Cluster size $n$', fontsize=15)
        ax.set_xticks([1,10,100,1000,3000])    
        ax.set_xlim(0, 3000)
        ax.set_ylim(0.00005, 2)
        ax.legend(fontsize=10, markerscale=1.5)
        ax.set_title('G3BP1 Cluster Size Distribution', pad=0, fontsize=15)
```

```python
scaling_factors = ['1.25']
s = 0

rna_names = ['rA40','shortA','longA']
rna_chain_lengths = [6, 20, 66]
rna_chain_nums = [2750, 825, 250]

max_range=150
minSize = 10
maxSize = 500

for rna in range(0,len(rna_names)):
    curr_system = rna_names[rna]
    save_name = scaling_factors[s]+'_'+curr_system+'_'+str(rna_chain_nums[rna])

    no_chain2 = rna_chain_nums[rna]
    no_chains = no_chain1 + no_chain2
    print(save_name)

    chain2_length = rna_chain_lengths[rna]
    chain1_bound=[0, chain1_length*no_chain1]
    chain2_bound=[chain1_bound[1], chain1_bound[1]+(chain2_length*int(no_chain2))]

    A0, A1 =[0, chain1_length*no_chain1]
    B0, B1 =[chain1_bound[1], chain1_bound[1]+(chain2_length*int(no_chain2))]
    
    all_coords = np.load('g3bp1_rna_data/'+save_name+'_all_coords.npy', allow_pickle=True)        
    clusterList = np.load('g3bp1_rna_data/'+save_name+'_cluster_ids.npy', allow_pickle=True)        
    clusterSizes = np.load('g3bp1_rna_data/'+save_name+'_cluster_sizes.npy', allow_pickle=True)   

    distances_from_com = [[] for i in range(0,2)] # 0 for G3BP1, 1 for RNA

    for snapshot in range(0,len(clusterSizes)):
        curr_sizeList = clusterSizes[snapshot]
        coord_list = all_coords[snapshot]

        for cc in range(0,len(clusterList[snapshot])):
            if len(clusterList[snapshot][cc]) <= maxSize and len(clusterList[snapshot][cc]) >= minSize:
                                    
                g3bp1_list = [chain for chain in clusterList[snapshot][cc] if 1 <= chain <= no_chain1]
                rna_list = [chain for chain in clusterList[snapshot][cc] if no_chain1+1 <= chain <= no_chains+1]

                g3bp1_chains = []
                for ele in g3bp1_list:
                    start = A0 + (ele-1)*chain1_length
                    end = start + chain1_length
                    g3bp1_chains.append(coord_list[start:end])
                g3bp1_points = np.array([c for chain in g3bp1_chains for c in chain])  

                rna_chains = []
                for ele in rna_list:
                    b_idx = ele - (no_chain1 + 1) 
                    start = B0 + b_idx*chain2_length
                    end = start + chain2_length
                    rna_chains.append(coord_list[start:end])
                rna_points = np.array([c for chain in rna_chains for c in chain])  
        
                all_points = np.concatenate((g3bp1_points, rna_points), axis=0)
                com=center_of_mass(all_points, [[0,120],[0,120],[0,120]]) 
        
                tmp_dist = []
                for b in range(0,len(g3bp1_points)):
                    tmp_dist.append(distance_ij(g3bp1_points[b],com,120))
                All_hist,r=np.histogram(tmp_dist,bins=radial_bins, density=True)
                distances_from_com[0].append(All_hist)    

                tmp_dist = []
                for b in range(0,len(rna_points)):
                    tmp_dist.append(distance_ij(rna_points[b],com,120))
                All_hist,r=np.histogram(tmp_dist,bins=radial_bins, density=True)
                distances_from_com[1].append(All_hist)    
                
    fig, ax = plt.subplots(figsize=(5,4),dpi=150)

    ax.plot(radial_bins[0:max_range], np.mean(distances_from_com[0], axis=0)[0:max_range], linestyle='-', color='mediumorchid', linewidth=1.5, label='G3BP1')
    ax.plot(radial_bins[0:max_range], np.mean(distances_from_com[1], axis=0)[0:max_range], linestyle='-', color=rna_colors[rna], linewidth=1.5, label=rna_label_names[rna])
    ax.legend(fontsize=10, loc='upper right')
        
    ax.set_xlabel('Distance from the COM (l.u)', fontsize=10)
    ax.set_ylabel('Density of Beads', fontsize=10)
    ax.axhline(0, linestyle='--', c='gray', linewidth=2, alpha=0.5)
    
    plt.show()
```
